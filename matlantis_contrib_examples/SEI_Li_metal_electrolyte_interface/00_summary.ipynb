# Reactive MD of a Li-metal / Carbonate-Electrolyte Interface

When lithium metal touches a carbonate electrolyte, solvent molecules and anions are reduced at its surface, and their
products form the solid electrolyte interphase (SEI). This example follows the first 100 ps of this process with molecular
dynamics. Because bond breaking is described without reaction-specific parameters, the decomposition pathways, the
electron transfer from Li and the composition of the forming interphase can be observed directly. The model is a Li(100)
slab in contact with LiPF<sub>6</sub> in EC/EMC (3:7 w/w) with a vinylene carbonate (VC) additive.

## Workflow

| Notebook | Description |
|---|---|
| [01_build_interface.ipynb](01_build_interface.ipynb) | Molecules from SMILES, optimized bcc Li lattice constant, Li(100) slab (216 atoms), Packmol packing of the electrolyte (864 atoms) around the slab and interface minimization with fixed slab (FIRE → LBFGS, up to 2,000 steps each) |
| [02_run_md.ipynb](02_run_md.ipynb) | NVT Langevin MD (100,000 steps, 300 K, 1 fs/step → 100 ps) with the central Li layers fixed; restartable, background job recommended |
| [03_analysis.ipynb](03_analysis.ipynb) | Charge transfer, automatic detection of decomposition events, reaction products, decomposition of selected molecules and near-surface elemental composition |
| [99_installation.ipynb](99_installation.ipynb) | Installation of `watermark` and Packmol (run first) |

## Expected Results

- **Interface** (01): A Li(100) slab between two electrolyte layers (1,080 atoms, two interfaces); molecules touching Li may already react during the minimization.
- **MD** (02): The temperature fluctuates around 300 K. The potential energy decreases in steps and gradually over the whole run (about 0.06 eV/atom between 5 and 100 ps with the defaults) as exothermic decompositions occur.
- **Charge transfer** (03): Electrons move from Li to the electrolyte; with the defaults the charge of the slab rises from +15.6 e to +64.3 e. Each decomposed molecule transfers about 4–5 e.
- **Decomposition** (03): 10 of 68 molecules decompose with the defaults (PF<sub>6</sub><sup>−</sup> 1/4, VC 0/4, EC 4/20, EMC 5/40). Molecules touching Li react within the first ~10 ps; molecules starting up to ~5 Å away react later (30–65 ps) after diffusing to the surface, and molecules further away stay intact. Each species follows characteristic pathways: EC → C<sub>2</sub>H<sub>4</sub>O<sub>2</sub> + CO, EMC → C<sub>2</sub>H<sub>5</sub>O + CH<sub>3</sub>O + CO (via CH<sub>3</sub>OCO), PF<sub>6</sub><sup>−</sup> → 6 F + P with F binding to Li and P entering the slab, and VC → C<sub>2</sub>H<sub>2</sub>O<sub>2</sub> + CO in runs where VC decomposes.
- **Near-surface composition** (03): F and P are enriched near Li (enrichment factors 1.62 and 1.67 with the defaults; F stays above the bulk value for cutoffs of 3–8 Å), through decomposition products and through intact PF<sub>6</sub><sup>−</sup> attracted to the positively charged surface.
- **Run-to-run variability**: In test runs differing only in the initial structure, the pathways, the F enrichment (1.5–1.95) and the charge transferred per decomposed molecule were reproduced, but the number of decomposed molecules (6–17) and the transferred charge (+38 to +86 e) varied strongly, and VC or PF<sub>6</sub><sup>−</sup> did not decompose at all in some runs. This is a statistical fluctuation, because there are only four PF<sub>6</sub><sup>−</sup> and four VC molecules and few molecules touch the surface. For quantitative amounts, use a larger system (a larger in-plane cell with more molecules) or average several runs with different initial structures.

**Estimated runtime:** about 3.5 h in total.
1,080 atoms (216 Li in the slab, 72 of them fixed, and 864 electrolyte atoms). 01: about 10–15 min (minimization up to 4,000 steps). 02: about 3–3.5 h (100,000 steps). 03: about 5–15 min (1,000 frames).

# Li 金属／カーボネート電解液界面の反応 MD

リチウム金属がカーボネート系電解液に触れると、表面で溶媒分子やアニオンが還元され、その生成物が固体電解質界面（SEI）を作ります。
この例では、その最初の 100 ps を分子動力学で追跡します。反応ごとのパラメータなしに結合の切断を扱えるため、分解経路、Li からの電子移動、
形成中の界面層の組成を直接観察できます。モデルは、Li(100) スラブと、ビニレンカーボネート（VC）を添加した LiPF<sub>6</sub> の EC/EMC（重量比 3:7）溶液が接した系です。

| [02_run_md.ipynb](02_run_md.ipynb) | 中央の Li 層を固定した NVT Langevin MD（100,000 ステップ、300 K、1 fs/ステップ → 100 ps）。再開可能、バックグラウンドジョブ推奨 |
| [03_analysis.ipynb](03_analysis.ipynb) | 電荷移動、分解イベントの自動検出、反応生成物、分子の分解過程、表面近傍の元素組成 |
| [99_installation.ipynb](99_installation.ipynb) | `watermark` と Packmol のインストール（最初に実行） |

## 期待される結果

- **界面**（01）：2 つの電解液層に挟まれた Li(100) スラブ（1,080 原子、界面 2 つ）。Li に接した分子が緩和の段階で反応が始まる可能性があります。
- **MD**（02）：温度は 300 K 付近で揺らぎます。発熱的な分解が起こるにつれ、ポテンシャルエネルギーが計算全体にわたって段階的・連続的に下がります。
- **電荷移動**（03）：Li から電解液へ電子が移ります。既定の設定では、スラブの電荷が +15.6 e から +64.3 e 程度まで増えます。分解 1 分子あたりの電荷移動は約 4〜5 e です。
- **分解**（03）：既定の設定では 68 分子中 10 分子程度が分解します（PF<sub>6</sub><sup>−</sup> 1/4、VC 0/4、EC 4/20、EMC 5/40）。Li に接していた分子は最初の約 10 ps で反応し、最大 5 Å 程度離れていた分子は表面まで拡散してから遅れて（30〜65 ps）反応し、それより遠い分子は分解しません。分子種ごとに特徴的な分解経路をとり、EC → C<sub>2</sub>H<sub>4</sub>O<sub>2</sub> + CO、EMC → C<sub>2</sub>H<sub>5</sub>O + CH<sub>3</sub>O + CO（CH<sub>3</sub>OCO を経由）、PF<sub>6</sub><sup>−</sup> → 6 F + P（F は Li と結びつき、P はスラブの中に入り込む）、VC が分解する計算では VC → C<sub>2</sub>H<sub>2</sub>O<sub>2</sub> + CO となります。
- **表面近傍の組成**（03）：Li の近くで F と P が濃縮されます（既定の設定で濃縮度 1.62 と 1.67。厚さ 3〜8 Å で F は常に電解液全体の値を上回ります）。分解生成物によるものと、正に帯電した表面に引き寄せられた、分解していない PF<sub>6</sub><sup>−</sup> によるものがあります。
- **計算ごとのばらつき**：初期構造だけを変えたテスト計算で、分解経路、F の濃縮（1.5〜1.95）、分解 1 分子あたりの電荷移動は再現しましたが、分解した分子の数（6〜17）と移った電荷（+38〜+86 e）は大きくばらつき、VC や PF<sub>6</sub><sup>−</sup> がまったく分解しない場合もあります。これは、PF<sub>6</sub><sup>−</sup> や VC が 4 個ずつしかなく、表面に接する分子の数も少ないことによる統計的なばらつきです。定量的に議論するには、系を大きくする（面内のサイズを広げて分子の数を増やす）か、初期構造を変えた計算を複数回行って平均してください。

**推定実行時間：** 合計約 3.5 時間。
1,080 原子（スラブの Li 216 原子、うち 72 原子を固定、電解液 864 原子）。01：約 10〜15 分（構造緩和は最大 4,000 ステップ）。02：約 3〜3.5 時間（100,000 ステップ）。03：約 5〜15 分（1,000 フレーム）。
