# Free Energy Calculation of Cu Elution from Cu(111)/Water Interface Using Umbrella Sampling

![Graphical Abstract](./assets/movies/traj-movie-small.gif)

This example demonstrates a workflow for calculating the free energy profile of pulling a Cu surface atom in the z-direction (perpendicular to the surface) at a **Cu(111) slab/water interface** using umbrella sampling. Specifically, the reaction coordinate space is explored using Steered MD with PLUMED's MOVINGRESTRAINT, umbrella sampling is performed at each window, and the free energy is calculated using the MBAR method.

## Notebook Overview

The calculation is divided into the following 6 steps (notebooks). By executing them in order, you can perform the entire workflow from modeling to analysis.
Input files required for each notebook are prepared in the `assets` directory, so each notebook can also be run independently.

| Step | Notebook | Overview |
| :--- | :--- | :--- |
| **01** | [01_modeling_cu_water_interface_en.ipynb](./01_modeling_cu_water_interface_en.ipynb) | **Modeling**<br>Create and optimize a Cu(111) slab and water molecule interface structure using ASE and pfcc_extras. |
| **02** | [02_equilibrium_npzt_md_en.ipynb](./02_equilibrium_npzt_md_en.ipynb) | **Equilibration MD**<br>Equilibrate the interface structure at 375 K using the NPzT ensemble (pressure control in z-direction only). |
| **03** | [03_steered_md_en.ipynb](./03_steered_md_en.ipynb) | **Steered MD (SMD)**<br>Use PLUMED's MOVINGRESTRAINT to pull the Cu surface atom in the z-direction and create a trajectory covering the entire reaction coordinate. |
| **04** | [04_select_umbrella_sampling_initial_structures_en.ipynb](./04_select_umbrella_sampling_initial_structures_en.ipynb) | **Initial Structure Selection**<br>Extract structures corresponding to each umbrella sampling window (specific reaction coordinate values) from the SMD trajectory. |
| **05** | [05_umbrella_sampling_en.ipynb](./05_umbrella_sampling_en.ipynb) | **Umbrella Sampling**<br>Using the extracted structures as starting points, run multiple MDs restrained by harmonic potentials in parallel with joblib to collect data. |
| **06** | [06_mbar_free_energy_en.ipynb](./06_mbar_free_energy_en.ipynb) | **Free Energy Analysis**<br>Analyze the obtained data using the MBAR method (pymbar) and plot the free energy profile (PMF) as a function of the Cu atom z-coordinate. |

## Target System and Reaction Coordinate

* **System**: Cu(111) slab/water interface
    * Cu(111) slab: 6×8×6 (288 atoms)
    * Water molecules: 250 (750 atoms)
    * Total: 1038 atoms, simulation under periodic boundary conditions
* **Reaction Coordinate (Collective Variable)**:
    * Z-coordinate of a single Cu surface atom (ASE index: 267, PLUMED index: 268)
    * This position is varied from approximately 10.2 Å to 18.0 Å.

## Required Libraries and Environment Setup

The following software and packages are required to run this example:

* **PFP-API-CLIENT**: Package for using PFP (Preferred Potential).
* **ASE (Atomic Simulation Environment)**: Used as the interface for atomic structure manipulation and MD.
* **pfcc_extras**: Used for LiquidGenerator (liquid structure generation), surface structure construction tools, etc.
* **PLUMED**: Used to apply harmonic restraints in Steered MD and Umbrella Sampling (Steps 03, 05).
    * Uses ASE's PLUMED Calculator interface.
* **pymbar**: Used for free energy analysis (MBAR method) in Step 06.
* **joblib**: Used for parallel execution of umbrella sampling in Step 05.

### How to Install PLUMED
[PLUMED](https://www.plumed.org/) is an open-source library that works with molecular dynamics simulation packages to calculate reaction coordinates (collective variables) and apply bias forces, enabling efficient sampling of rare events and free energy analysis. In this notebook, PLUMED is used to add harmonic restraints to the reaction coordinate (Cu atom z-coordinate), so it must be installed beforehand.

The PLUMED installation procedure is as follows:

#### 1. Download and Compile the Source Code
Run the following commands in the terminal to build PLUMED:

```bash

# Create installation directory
mkdir -p ~/local && cd ~/local

# Clone the PLUMED repository
git clone https://github.com/plumed/plumed2.git plumed-2.9.0

# Checkout version v2.9.0
cd plumed-2.9.0 && git checkout v2.9.0

# Build PLUMED (configure & make)
./configure --disable-mpi --prefix=$HOME/local/plumed-2.9.0 && make -j\"$(nproc)\" && make install
```

#### 2. Install Python Bindings
Install the Python package for calling PLUMED from ASE:

```bash
$ pip install plumed
```
