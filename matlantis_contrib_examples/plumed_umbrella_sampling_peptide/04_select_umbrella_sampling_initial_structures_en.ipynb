# Selecting Initial Structures for Umbrella Sampling

In the previous steps, we obtained data from stretching Deca-alanine (Steered MD trajectories).
The subsequent Umbrella Sampling involves running numerous independent simulations centered at specific distances (e.g., $d = 13.0$ Å, $13.5$ Å, ...).
In this notebook, we will extract the snapshots from the Steered MD trajectory data that are closest to the target distance for each umbrella window and use them as initial structures.

## Step 1. Import Libraries

```python
import os
import numpy as np
import matplotlib.pyplot as plt

from ase.io import write, Trajectory
```

## Step 2. Loading Structure Files

```python
# Directory for trajectory files
traj_dir = "./output/03_steered_md"

# Output directory
out_dir  = "./input/04_umbrella_sampling"
os.makedirs(out_dir, exist_ok=True)

# Load reaction coordinate trajectories
cv_traj = []
cv_range = np.arange(13.0, 33.1, 0.5) # Set the range and step size for CV

for cv in cv_range:
    cv_str = f'{cv:.2f}'
    cv_traj.append(np.loadtxt(f"{traj_dir}/cv_{cv_str}/COLVAR_{cv_str}")[1:, 1])

cv_traj = np.array(cv_traj)
cv_traj.shape
```

## Step 3. Extract Initial Structures for Each Umbrella Window

For each target distance (CV), we load the corresponding Steered MD trajectory file and select the snapshot that has the smallest difference from the target distance.

```python
# Prepare Initial Structures

print(f"  i,     cv,  cv of struct.")
for i, cv in enumerate(cv_range):
    diff = np.abs(cv - cv_traj[i])
    jmin = np.argmin(diff)
    print(f"{i:3d},  {cv:.2f},  {cv_traj[i][jmin]:.2f}")

    cv_str = f'{cv:.2f}'
    atoms = Trajectory(f'{traj_dir}/cv_{cv_str}/md-dyn.traj') 
    write(f'{out_dir}/initial_cv_{cv:.2f}.xyz', atoms[jmin])
```
